```python
import random

def f(x):
    return -(x*x) + (6*x)


def hill_climbing():

    min_x = 0
    max_x = 6

    curr_x = random.randint(min_x, max_x)
    print("Initial value of x:", curr_x)

    step = 1

    while True:

        curr_f = f(curr_x)
        print("Step", step, "-> Current value of x:", curr_x, "| f(x):", curr_f)

        neighbors = []

        if curr_x + 1 <= max_x:
            neighbors.append(curr_x + 1)

        if curr_x - 1 >= min_x:
            neighbors.append(curr_x - 1)

        best_neighbor = None
        best_neighbor_f = curr_f

        for n in neighbors:

            n_f = f(n)

            if n_f > best_neighbor_f:
                best_neighbor = n
                best_neighbor_f = n_f

        if best_neighbor == None:

            print("Final optimal value of x:", curr_x)
            print("Maximum f(x):", curr_f)
            break

        curr_x = best_neighbor
        step = step + 1


hill_climbing()
```

```python
import heapq

def heuristic(x):
    return abs(20 - x)


def beam_search(start, goal, width = 2):

    beam = [(heuristic(start), [start])]
    level = 0

    while beam:

        print("Level", level)

        candidates = []

        for h, path in beam:

            x = path[-1]
            print("Number explored:", x)

            if x == goal:
                print("\nGoal has been reached!")
                return path

            for n in [x + 2, x + 3, x * 2]:

                new_path = path + [n]
                candidates.append((heuristic(n), new_path))

        beam = heapq.nsmallest(width, candidates, key = lambda x: x[0])

        print("Beam states:")

        for h, p in beam:
            print(p[-1], "heuristic:", h)


        level = level + 1


result = beam_search(1, 20)

print("Final path:", result)
```

```python
import random

POPULATION_SIZE = 6
GENERATIONS = 15
MUTATION_RATE = 0.1


def fitness(x):
    return x*x + 2*x


def binary_to_decimal(b):
    return int(b, 2)


def create_chromosome():
    return format(random.randint(0, 31), '05b')


def initialize_population():

    pop = []

    for i in range(POPULATION_SIZE):
        pop.append(create_chromosome())

    return pop


def selection(pop):

    pop = sorted(pop, key = lambda c: fitness(binary_to_decimal(c)), reverse = True)

    return pop[0], pop[1]


def crossover(p1, p2):

    point = random.randint(1, 4)

    child1 = p1[:point] + p2[point:]
    child2 = p2[:point] + p1[point:]

    return child1, child2


def mutation(ch):

    if random.random() < MUTATION_RATE:

        pos = random.randint(0, 4)

        ch_list = list(ch)

        if ch_list[pos] == '0':
            ch_list[pos] = '1'
        else:
            ch_list[pos] = '0'

        ch = "".join(ch_list)

    return ch


def genetic_algorithm():

    population = initialize_population()

    for gen in range(GENERATIONS):

        p1, p2 = selection(population)

        c1, c2 = crossover(p1, p2)

        c1 = mutation(c1)
        c2 = mutation(c2)

        population[-1] = c1
        population[-2] = c2

    best = max(population, key = lambda c: fitness(binary_to_decimal(c)))

    return best


best_ch = genetic_algorithm()

best_x = binary_to_decimal(best_ch)

print("Best Chromosome:", best_ch)
print("Best value of x:", best_x)
print("Best Fitness:", fitness(best_x))
```
