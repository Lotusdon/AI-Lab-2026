```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error


data = pd.read_csv("/kaggle/input/house-prices-advanced-regression-techniques/train.csv")


data.drop("Id", axis = 1, inplace = True)


data.fillna(data.median(numeric_only = True), inplace = True)
data.fillna("None", inplace = True)


data = pd.get_dummies(data)


x = data.drop("SalePrice", axis = 1)
y = data["SalePrice"]


xtrain, xtest, ytrain, ytest = train_test_split(x, y,
                                                test_size = 0.2)


scaler = StandardScaler()

xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)


lin = LinearRegression()
tree = DecisionTreeRegressor()


lin.fit(xtrain, ytrain)
tree.fit(xtrain, ytrain)


lin_pred = lin.predict(xtest)
tree_pred = tree.predict(xtest)


lin_mae = mean_absolute_error(ytest, lin_pred)
lin_rmse = np.sqrt(mean_squared_error(ytest, lin_pred))


print("\n== linear regression")

print("\tmae :", lin_mae)
print("\trmse :", lin_rmse)


tree_mae = mean_absolute_error(ytest, tree_pred)
tree_rmse = np.sqrt(mean_squared_error(ytest, tree_pred))


print("\n== decision tree")

print("\tmae :", tree_mae)
print("\trmse :", tree_rmse)
```
